```python
import pandas as pd

df = pd.read_csv("urniki.csv")

letos = df[
    df["semester"] == "Poletni semester 2025/26"
]

letos["utezene ure"] = letos["trajanje"] / letos["stevilo programov"]

letos.groupby("predavalnica")["utezene ure"].sum().sort_values(ascending=False).head(20)
```

```python
df["predavatelj"].iloc(0)
predavatelji = df.explode("predavatelj")

ure = (
    predavatelji
    .groupby("predavatelj")["utezene ure"]
    .sum()
    .sort_values(ascending=False)
    / df["semester"].nunique()
)
ure.head(30)
```
